```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the CSV
DATA_PATH = "../data/raw/CommunityHealthcare_MultiSymptomsDisease_Diagnostic_Dataset.csv"
```

```python
df = pd.read_csv(DATA_PATH)
print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Inspect the dataset structure
print("Number of records:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes.value_counts())

print("\nDataset information:")
df.info()
```

```python
# Check missing values and duplicates
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nColumns with missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nUnique values in each column:")
print(df.nunique().sort_values())
```

```python
# Analyze the target variable
if "disease" not in df.columns:
    raise ValueError("Target column 'disease' was not found.")

disease_counts = df["disease"].value_counts()

print("Number of disease classes:", df["disease"].nunique())
print("\nRecords per disease:")
print(disease_counts)

print("\nLeast represented classes:")
print(disease_counts.sort_values().head(10))
```

```python
# Visualize 15 most common disease labels
top_diseases = df["disease"].value_counts().head(15)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_diseases.values,
    y=top_diseases.index
)

plt.title("Top 15 Disease Categories")
plt.xlabel("Number of Records")
plt.ylabel("Disease")
plt.tight_layout()
plt.show()
```
