```python
# consolidated comparison report
import pandas as pd
from pathlib import Path

REPORT_DIR = Path("../reports")

rf = pd.read_csv(
    REPORT_DIR / "random_forest_metrics.csv"
).iloc[0]

xgb = pd.read_csv(
    REPORT_DIR / "xgboost_metrics.csv"
).iloc[0]

cat = pd.read_csv(
    REPORT_DIR / "catboost_metrics.csv"
).iloc[0]

comparison = pd.DataFrame([
    {
        "Model": "Random Forest",
        "Accuracy": rf["Accuracy"],
        "Macro F1": rf["Macro F1"],
        "Weighted F1": rf["Weighted F1"]
    },
    {
        "Model": "XGBoost",
        "Accuracy": xgb["Accuracy"],
        "Macro F1": xgb["Macro F1"],
        "Weighted F1": xgb["Weighted F1"]
    },
    {
        "Model": "CatBoost",
        "Accuracy": cat["Accuracy"],
        "Macro F1": cat["Macro F1"],
        "Weighted F1": cat["Weighted F1"]
    }
])

comparison = comparison.sort_values(
    "Macro F1", ascending=False
)

display(comparison.round(4))

comparison.to_csv(
    REPORT_DIR / "model_comparison.csv",
    index=False
)

print("Comparison report saved.")
```
